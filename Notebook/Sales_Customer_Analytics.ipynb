Stage 3: Data Transformation

Part A: Sales Calculations
1. TotalSales naam ka column create karo. Formula: Quantity × UnitPrice.
2. DiscountAmount column create karo. Formula: TotalSales × Discount.
3. NetSales column create karo. Formula: TotalSales - DiscountAmount.
4. Profit naam ka column create karo. Assume karo company ka profit margin NetSales ka 20% hai.

```python
df["TotalSales"] = df["Quantity"]*df["UnitPrice"]
```

```python
df["DiscountAmount"] = df["TotalSales"]*df["Discount"]
```

```python
df["NetSales"] = df["TotalSales"] - df["DiscountAmount"]
```

```python
df["Profit"] = df["NetSales"]*0.2
```

Part B: Date Transformation
1. OrderDate se OrderYear column create karo.
6. OrderDate se OrderMonth column create karo.
7. OrderDate se OrderDay column create karo.
8. OrderDate se OrderQuarter column create karo.

```python
df["OrderYear"] = df["OrderDate"].dt.year
```

```python
df["OrderMonth"] = df["OrderDate"].dt.month
```

```python
df["OrderDay"] = df["OrderDate"].dt.day
```

```python
df["OrderQuarter"] = df["OrderDate"].dt.quarter
```

Part C: Business Categorization
1. NetSales ke basis par SalesCategory create karo:
- NetSales >= 50000 → High
- NetSales >= 20000 → Medium
- NetSales < 20000 → Low
2. Quantity ke basis par OrderType create karo:
- Quantity >= 3 → Bulk Order
- Quantity < 3 → Regular Order

```python
df["SalesCategory"] = df["NetSales"].apply(
    lambda x: "High" if x >= 50000 else("Medium" if x>= 20000 else "Low")
)
print(df)
```

```python
df["OrderType"] = df["Quantity"].apply(
    lambda x : "Bulk Order" if x >= 3 else"Regular Order"
)
```
